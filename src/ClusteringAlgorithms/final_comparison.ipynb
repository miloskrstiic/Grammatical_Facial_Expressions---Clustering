```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

RESULTS_ROOT = "../../results"
FINAL_DIR = "../../results/final_comparison"

os.makedirs(FINAL_DIR, exist_ok=True)
```

```python
kmeans = pd.read_csv(f"{RESULTS_ROOT}/kmeans/kmeans_results_summary.csv")
dbscan = pd.read_csv(f"{RESULTS_ROOT}/dbscan/dbscan_grid_summary.csv")
hdbscan = pd.read_csv(f"{RESULTS_ROOT}/hdbscan/hdbscan_results_summary.csv")
ahc = pd.read_csv(f"{RESULTS_ROOT}/ahc/ahc_best_results.csv")
gmm = pd.read_csv(f"{RESULTS_ROOT}/gmm/gmm_best_results.csv")
```

### Standardizacija KMeans rezultata

```python
kmeans_final = kmeans.copy()

kmeans_final = kmeans_final.rename(columns={
    "K": "Clusters"
})

kmeans_final["Noise_ratio"] = np.nan
kmeans_final["Selection_score"] = np.nan
kmeans_final["BIC"] = np.nan
kmeans_final["AIC"] = np.nan

kmeans_final = kmeans_final[
    [
        "Dataset",
        "Algorithm",
        "Clusters",
        "Silhouette",
        "Davies_Bouldin",
        "ARI",
        "NMI",
        "Noise_ratio",
        "Selection_score",
        "BIC",
        "AIC"
    ]
]

kmeans_final
```

### Standardizacija DBSCAN rezultata

```python
dbscan_final = dbscan.copy()

dbscan_final = dbscan_final.rename(columns={
    "Silhouette_non_noise": "Silhouette",
    "Davies_Bouldin_non_noise": "Davies_Bouldin"
})

dbscan_final["BIC"] = np.nan
dbscan_final["AIC"] = np.nan

dbscan_final = dbscan_final[
    [
        "Dataset",
        "Algorithm",
        "Clusters",
        "Silhouette",
        "Davies_Bouldin",
        "ARI",
        "NMI",
        "Noise_ratio",
        "Selection_score",
        "BIC",
        "AIC"
    ]
]

dbscan_final
```

### Standardizacija HDBSCAN rezultata

```python
hdbscan_final = hdbscan.copy()

hdbscan_final = hdbscan_final.rename(columns={
    "Silhouette_non_noise": "Silhouette",
    "Davies_Bouldin_non_noise": "Davies_Bouldin"
})

hdbscan_final["BIC"] = np.nan
hdbscan_final["AIC"] = np.nan

hdbscan_final = hdbscan_final[
    [
        "Dataset",
        "Algorithm",
        "Clusters",
        "Silhouette",
        "Davies_Bouldin",
        "ARI",
        "NMI",
        "Noise_ratio",
        "Selection_score",
        "BIC",
        "AIC"
    ]
]

hdbscan_final
```

### Standardizacija AHC rezultata

```python
ahc_final = ahc.copy()

ahc_final = ahc_final.rename(columns={
    "K": "Clusters"
})

ahc_final["Algorithm"] = "AHC"
ahc_final["Noise_ratio"] = np.nan
ahc_final["Selection_score"] = np.nan
ahc_final["BIC"] = np.nan
ahc_final["AIC"] = np.nan

ahc_final = ahc_final[
    [
        "Dataset",
        "Algorithm",
        "Clusters",
        "Silhouette",
        "Davies_Bouldin",
        "ARI",
        "NMI",
        "Noise_ratio",
        "Selection_score",
        "BIC",
        "AIC"
    ]
]

ahc_final
```

### Standardizacija GMM rezultata

```python
gmm_final = gmm.copy()

gmm_final = gmm_final.rename(columns={
    "K": "Clusters"
})

gmm_final["Noise_ratio"] = np.nan
gmm_final["Selection_score"] = np.nan

gmm_final = gmm_final[
    [
        "Dataset",
        "Algorithm",
        "Clusters",
        "Silhouette",
        "Davies_Bouldin",
        "ARI",
        "NMI",
        "Noise_ratio",
        "Selection_score",
        "BIC",
        "AIC"
    ]
]

gmm_final
```

# Spajanje svih rezultata

```python
final_comparison = pd.concat(
    [
        kmeans_final,
        dbscan_final,
        hdbscan_final,
        ahc_final,
        gmm_final
    ],
    ignore_index=True
)

final_comparison = final_comparison.round({
    "Silhouette": 4,
    "Davies_Bouldin": 4,
    "ARI": 4,
    "NMI": 4,
    "Noise_ratio": 4,
    "Selection_score": 4,
    "BIC": 2,
    "AIC": 2
})

final_comparison.to_csv(
    f"{FINAL_DIR}/final_comparison_summary.csv",
    index=False
)

final_comparison
```

### Najbolji algoritmi po metrikama

```python
best_by_dataset = []

for dataset_name, group in final_comparison.groupby("Dataset"):
    row = {
        "Dataset": dataset_name
    }

    best_silhouette = group.loc[group["Silhouette"].idxmax()]
    row["Best_by_Silhouette"] = best_silhouette["Algorithm"]
    row["Best_Silhouette"] = best_silhouette["Silhouette"]

    best_davies = group.loc[group["Davies_Bouldin"].idxmin()]
    row["Best_by_Davies_Bouldin"] = best_davies["Algorithm"]
    row["Best_Davies_Bouldin"] = best_davies["Davies_Bouldin"]

    best_ari = group.loc[group["ARI"].idxmax()]
    row["Best_by_ARI"] = best_ari["Algorithm"]
    row["Best_ARI"] = best_ari["ARI"]

    best_nmi = group.loc[group["NMI"].idxmax()]
    row["Best_by_NMI"] = best_nmi["Algorithm"]
    row["Best_NMI"] = best_nmi["NMI"]

    best_by_dataset.append(row)

best_by_dataset = pd.DataFrame(best_by_dataset)

best_by_dataset.to_csv(
    f"{FINAL_DIR}/best_algorithm_by_dataset.csv",
    index=False
)

best_by_dataset
```

### Grafici za Silhouette, ARI i NMI

```python
metrics_to_plot = ["Silhouette", "ARI", "NMI"]

for metric in metrics_to_plot:
    plt.figure(figsize=(10, 5))

    sns.barplot(
        data=final_comparison,
        x="Algorithm",
        y=metric,
        hue="Dataset"
    )

    plt.title(f"Final comparison by {metric}")
    plt.xlabel("Algorithm")
    plt.ylabel(metric)
    plt.xticks(rotation=30)
    plt.tight_layout()

    filename = f"{FINAL_DIR}/final_comparison_{metric.lower()}.png"
    plt.savefig(filename, dpi=200)
    plt.close()

    print(f"Saved: {filename}")
```

### The final comparison combines the best configuration of each clustering algorithm for both datasets. Internal metrics such as Silhouette and Davies-Bouldin evaluate the geometric structure of the clusters, while ARI and NMI compare the obtained clusters with the true expression labels.

### For DBSCAN and HDBSCAN, Silhouette and Davies-Bouldin were computed only on non-noise points, because noise points are not regular clusters. Their Noise ratio and Selection score were therefore included separately.

### The results show that different algorithms optimize different aspects of the problem. Some methods produce better internal cluster separation, while others show slightly better agreement with the true expression labels. Overall, the low ARI and NMI values indicate that the expression classes are not clearly separated in the selected feature space.
